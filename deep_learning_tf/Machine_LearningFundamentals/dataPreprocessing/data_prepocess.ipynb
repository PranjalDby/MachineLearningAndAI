```python
import string

alphabet_list = list(string.ascii_lowercase) + list(string.ascii_uppercase)
print(alphabet_list)
```

```python
import random
##Shuffling the alphabet-list
random.shuffle(alphabet_list)
```

```python
new_ent = dict()
# Generate a list of departments
departments = ['HR', 'Finance', 'Engineering', 'Marketing', 'Sales', 'IT', 'Operations', 'Legal', 'R&D', 'Support']

for cols in df.columns.map(lambda x:x.strip()).to_list():
    new_ent[cols] = []
    for i in range(100):
        if cols == 'Name':
            string = ""
            for j in range(4):
                string += random.choice(alphabet_list)
        
            new_ent['Name'].append(string)
        elif cols == 'Age':
            new_ent[cols].append(random.randint(20,100))

        elif cols == "Department":
            new_ent[cols].append(random.choice(departments))
        
        elif cols == "Salary":
            new_ent[cols].append(float(random.randrange(20000,100000,5000)))

print(new_ent)
```

```python
df.columns = df.columns.map(lambda s:s.strip()).to_list()
new_df = pd.DataFrame(new_ent)
for i in range(5):
    cols = random.choice(new_df.columns)
    #choosing value
    index = random.randrange(0,len(new_df[cols]))
    new_df.loc[index,cols] = np.nan
    #putting nA there


# df.add(new_ent)
print(new_df.isna().sum())
print(new_df.to_string())
```

```python
# concatinating to our existing df
```

```python
updated_df = pd.concat([df,new_df],axis=0)

print(updated_df)
```

**Forward fill** and **backward fill** are techniques used to handle missing data in a DataFrame:

- **Forward fill (`ffill`)**: This method propagates the last valid observation forward to fill missing values. For example, if a value is missing, it will be replaced by the previous non-missing value in the column.
- **Backward fill (`bfill`)**: This method uses the next valid observation to fill missing values. If a value is missing, it will be replaced by the next non-missing value in the column.

These methods are commonly used with `pandas.DataFrame.fillna(method='ffill')` or `pandas.DataFrame.fillna(method='bfill')`.
